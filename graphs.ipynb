```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("train_preprocessed.csv")
```

```python
survival_rates = df.groupby('Sex')['Survived'].mean() * 100
survival_rates.index = survival_rates.index.map({0: 'Male', 1: 'Female'})
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(x=survival_rates.index, y=survival_rates.values,
            palette=['steelblue', 'salmon'])
plt.title('Survival Rate by Gender', fontsize=16)
plt.xlabel('Gender', fontsize=14)
plt.ylabel('Survival Rate (%)', fontsize=14)
for i, v in enumerate(survival_rates.values):
    plt.text(i, v + 1, f"{v:.1f}%", ha='center', fontsize=12)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
class_survival = df.groupby('Pclass')['Survived'].mean() * 100
sns.barplot(x=class_survival.index, y=class_survival.values, palette='viridis')
plt.title('Survival Rate by Passenger Class', fontsize=16)
plt.xlabel('Passenger Class', fontsize=14)
plt.ylabel('Survival Rate (%)', fontsize=14)
for i, v in enumerate(class_survival.values):
    plt.text(i, v + 1, f"{v:.1f}%", ha='center', fontsize=12)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='Age', hue='Survived', kde=True, element='step',
             stat='density', common_norm=False, palette=['salmon', 'steelblue'])
plt.title('Age Distribution: Survived vs. Not Survived', fontsize=16)
plt.xlabel('Age', fontsize=14)
plt.ylabel('Density', fontsize=14)
plt.legend(labels=['Did Not Survive', 'Survived'])
plt.tight_layout()
plt.show()
```
